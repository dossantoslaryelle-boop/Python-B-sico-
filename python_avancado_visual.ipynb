# 🚀 Python Avançado — Visualizações e Projetos

Este notebook mostra exemplos mais avançados com foco em gráficos, análise de dados e visualizações.

## 📦 Imports

```python
import pandas as pd
import matplotlib.pyplot as plt
import random
```

## 📊 1. Criando um dataset

```python
dados = {
    'Mes': ['Jan', 'Fev', 'Mar', 'Abr', 'Mai'],
    'Vendas': [100, 150, 200, 180, 220]
}

df = pd.DataFrame(dados)
df
```

## 📈 2. Gráfico de linha

```python
plt.figure()
plt.plot(df['Mes'], df['Vendas'])
plt.title('Vendas por Mês')
plt.xlabel('Mês')
plt.ylabel('Vendas')
plt.show()
```

## 📊 3. Gráfico de barras

```python
plt.figure()
plt.bar(df['Mes'], df['Vendas'])
plt.title('Comparação de Vendas')
plt.show()
```

## 🔵 4. Gráfico de dispersão

```python
x = [random.randint(0, 50) for _ in range(50)]
y = [random.randint(0, 50) for _ in range(50)]

plt.figure()
plt.scatter(x, y)
plt.title('Dispersão de Dados')
plt.show()
```

## 📉 5. Histograma

```python
dados = [random.gauss(50, 15) for _ in range(1000)]

plt.figure()
plt.hist(dados, bins=30)
plt.title('Distribuição dos Dados')
plt.show()
```

## 🎞️ 6. Simulação visual (loop)

```python
import time

for i in range(5):
    print('Carregando' + '.' * i)
    time.sleep(0.5)
```

## 📊 7. Análise de dados

```python
print('Média:', df['Vendas'].mean())
print('Máximo:', df['Vendas'].max())
print('Mínimo:', df['Vendas'].min())
```

## 🧠 8. Mini projeto — Dashboard simples

```python
plt.figure()
plt.plot(df['Mes'], df['Vendas'])
plt.bar(df['Mes'], df['Vendas'])
plt.title('Visão geral de vendas')
plt.show()
```
